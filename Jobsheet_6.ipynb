### 1.Melakukan Visualisasi Data menggunakan Pie Chart

```python
import matplotlib.pyplot as plt
flavors = ('Chocolate', 'Vanilla', 'Pistachio', 'Mango', 'Strawberry')
votes = (12, 11, 4, 8, 7)
colors = ('#8B4513', '#FFF8DC', '#93C572', '#E67F0D','#D53032')
explode = (0, 0, 0, 0.1, 0)
plt.pie(votes,
        labels=flavors,
       autopct='%1.2f%%',
       colors=colors,
       explode = explode,
       shadow = True)
plt.show()
```

### 2.Melakukan Visualisasi Data menggunakan Bar Chart

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
countries = ('Argentina', 'Bolivia', 'Brazil', 'Chile',  'Colombia', 'Ecuador', 'Falkland Islands', 'French Guiana', 'Guyana', 'Paraguay', 'Peru','Suriname', 'Uruguaiy', 'Venezuela')
populations = (45076704, 11626410, 212162757, 19109629,50819826, 17579085, 3481, 287750, 785409, 7107305, 32880332,585169, 3470475, 28258770)
df = pd.DataFrame({
    'Country' : countries,
    'Population' : populations,
})
df.sort_values(by='Population', inplace = True)
x_coords = np.arange(len(countries))
plt.bar(x_coords, df['Population'], tick_label=df['Country'])
plt.xticks(rotation=90) #rotates text for x-axis labels
plt.ylabel('Population')
plt.title('South American Populations')
plt.show()

```

### Latihan 2: Permen

```python
import matplotlib.pyplot as plt

# Data
candy_names = ['Kit Kat', 'Snickers', 'Milky Way', 'Toblerone', 'Twix']
candy_counts = [52, 39, 78, 13, 78]

# Hitung total permen
total_candies = sum(candy_counts)

# Hitung peluang memilih Snickers
snickers_index = candy_names.index('Snickers')
snickers_count = candy_counts[snickers_index]
snickers_probability = snickers_count / total_candies

# Cetak peluang
print(f"Peluang memilih Snickers: {snickers_probability:.2%}")

# Buat diagram pie
plt.figure(figsize=(8, 6))
plt.pie(candy_counts, labels=candy_names, autopct='%1.1f%%', startangle=140)
plt.title('Persentase Setiap Jenis Permen dalam Kantong Mario')
plt.axis('equal')  # Membuat pie jadi lingkaran sempurna
plt.show()
```

### Latihan 3: Makanan

```python
import matplotlib.pyplot as plt

# Data penjualan makanan penutup
dessert_sales = {
    'Lava Cake': 14,
    'Mousse': 5,
    'Chocolate Cake': 12,
    'Ice Cream': 19,
    'Truffles': 6,
    'Brownie': 8,
    'Chocolate Chip Cookie': 12,
    'Chocolate Pudding': 9,
    'Souffle': 10,
    'Chocolate Cheesecake': 17,
    'Chocolate Chips': 2,
    'Fudge': 9,
    'Mochi': 13,
}

# Mengurutkan berdasarkan jumlah penjualan (ascending)
sorted_sales = dict(sorted(dessert_sales.items(), key=lambda item: item[1]))

# Tiga makanan penutup dengan penjualan terendah
least_popular = list(sorted_sales.items())[:3]

# Tampilkan tiga yang akan disingkirkan
print("Tiga makanan penutup yang harus disingkirkan:")
for dessert, sales in least_popular:
    print(f"- {dessert} ({sales} penjualan)")

# Visualisasi diagram batang
plt.figure(figsize=(12, 6))
plt.bar(sorted_sales.keys(), sorted_sales.values(), color='chocolate')
plt.xticks(rotation=45)
plt.ylabel('Jumlah Penjualan')
plt.title('Popularitas Makanan Penutup (minggu terakhir)')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
