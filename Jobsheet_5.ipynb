### 1.Hapus Data yang kosong (NaN)

```python
df = pd.read_csv('./Data.csv')
df.dropna(inplace=True)

df
```

### 2.Ganti Data yang kosong (NaN) menjadi :

### <ul><li>Umur = 50 </li><li>Gaji = 25000000</li></ul>

```python
df = pd.read_csv('./Data.csv')
df['Umur'].fillna(50, inplace=True)
df['Gaji'].fillna(25000000, inplace=True)
df
```

### 3.Ganti Data Kosong (NaN) pada Umur dan Gaji dengan Nilai Rata-Rata

```python
df = pd.read_csv('./Data.csv')
mean_umur = df['Umur'].mean();
mean_gaji = df['Gaji'].mean();

df['Umur'].fillna(mean_umur, inplace=True)
df['Gaji'].fillna(mean_gaji, inplace=True)
df
```

### 4.Ganti Data Kosong (NaN) pada Umur dan Gaji dengan Nilai Median

```python
df = pd.read_csv('./Data.csv')
mean_umur = df['Umur'].median();
mean_gaji = df['Gaji'].median();

df['Umur'].fillna(mean_umur, inplace=True)
df['Gaji'].fillna(mean_gaji, inplace=True)

df
```

### 5. Ganti Data Kosong (NaN) pada Umur dan Gaji dengan Nilai Modus

```python
df = pd.read_csv('Data.csv')
mean_umur = df['Umur'].mode()[0];
mean_gaji = df['Gaji'].mode()[0];

df['Umur'].fillna(mean_umur, inplace=True)
df['Gaji'].fillna(mean_gaji, inplace=True)

df
```

### 6. Apabila Ada Umur yang diatas 120 maka ubah menjadi 33

```python
df = pd.read_csv('./Data.csv')

for x in df.index:
    if df.loc[x, 'Umur'] > 120:
        df.loc[x, 'Umur'] = 33

df
```

### 1.Buatlah table list Top 5 Customers yang memiliki nilai transaksi paling banyak !

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set style untuk visualisasi
sns.set(style="whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)

def load_and_prepare_data():
    """Memuat dan mempersiapkan data dari file CSV"""
    try:
        # Memuat data
        customers_df = pd.read_csv('customers.csv')
        transactions_df = pd.read_csv('transactions_1k.csv')

        # Persiapan data transaksi
        transactions_df['DateTime'] = pd.to_datetime(transactions_df['Date'] + ' ' + transactions_df['Time'])
        transactions_df['Hour'] = transactions_df['DateTime'].dt.hour

        # Gabungkan data pelanggan dengan transaksi
        customer_transactions = transactions_df.groupby('CustomerID').agg(
            TotalTransactions=('TransactionID', 'count'),
            TotalAmount=('Amount', 'sum'),
            TotalSpent=('Price', 'sum')
        ).reset_index()

        full_customer_data = customers_df.merge(customer_transactions, on='CustomerID', how='left')

        return full_customer_data, transactions_df
    except FileNotFoundError as e:
        print(f"Error: File tidak ditemukan - {e}")
        return None, None
    except Exception as e:
        print(f"Error: {e}")
        return None, None

def analyze_customers(customers_df):
    """Analisis data pelanggan"""
    print("\n=== ANALISIS PELANGGAN ===")

    # Top 5 pelanggan berdasarkan jumlah transaksi
    top_by_transactions = customers_df.sort_values('TotalTransactions', ascending=False).head(5)
    print("\nTop 5 Pelanggan berdasarkan Jumlah Transaksi:")
    print(top_by_transactions[['CustomerID', 'Segment', 'TotalTransactions']].to_string(index=False))

    # Top 5 pelanggan berdasarkan total pengeluaran
    top_by_spending = customers_df.sort_values('TotalSpent', ascending=False).head(5)
    print("\nTop 5 Pelanggan berdasarkan Total Pengeluaran:")
    print(top_by_spending[['CustomerID', 'Segment', 'TotalSpent']].to_string(index=False))

    # Simpan hasil analisis
    top_by_transactions.to_csv('top_customers_by_transactions.csv', index=False)
    top_by_spending.to_csv('top_customers_by_spending.csv', index=False)

def analyze_transactions(transactions_df):
    """Analisis data transaksi - hanya menampilkan 5 transaksi terbanyak"""
    print("\n=== ANALISIS 5 TRANSAKSI TERBANYAK ===")

    # Ambil 5 transaksi dengan Amount terbesar
    top_transactions = transactions_df.nlargest(5, 'Amount')

    print("\n5 Transaksi dengan Jumlah Pembelian Terbanyak:")
    print(top_transactions[['TransactionID', 'CustomerID', 'ProductID', 'Amount', 'Price', 'DateTime']].to_string(index=False))

    # Visualisasi
    plt.figure(figsize=(12, 6))
    sns.barplot(x='TransactionID', y='Amount', data=top_transactions, palette='viridis')
    plt.title('5 Transaksi dengan Jumlah Pembelian Terbanyak')
    plt.xlabel('ID Transaksi')
    plt.ylabel('Jumlah Pembelian (liter)')
    plt.xticks(rotation=45)
    plt.show()

def main():
    """Fungsi utama untuk menjalankan analisis"""
    print("Memulai analisis data pelanggan dan transaksi...")

    # Memuat data
    customers_df, transactions_df = load_and_prepare_data()

    if customers_df is not None and transactions_df is not None:
        # Jalankan analisis
        analyze_customers(customers_df)
        analyze_transactions(transactions_df)

    print("\nAnalisis selesai. Hasil telah disimpan dalam file CSV.")

if __name__ == "__main__":
    main()
```
