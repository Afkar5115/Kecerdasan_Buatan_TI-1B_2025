### 1.Hapus Data yang kosong (NaN)

### 2.Ganti Data yang kosong (NaN) menjadi :

### <ul><li>Umur = 50 </li><li>Gaji = 25000000</li></ul>

```python
df = pd.read_csv('./Data.csv')
df['Umur'].fillna(50, inplace=True)
df['Gaji'].fillna(25000000, inplace=True)
df
```

### 3.Ganti Data Kosong (NaN) pada Umur dan Gaji dengan Nilai Rata-Rata

```python
df = pd.read_csv('./Data.csv')
mean_umur = df['Umur'].mean();
mean_gaji = df['Gaji'].mean();

df['Umur'].fillna(mean_umur, inplace=True)
df['Gaji'].fillna(mean_gaji, inplace=True)
df
```

### 4.Ganti Data Kosong (NaN) pada Umur dan Gaji dengan Nilai Median

```python
df = pd.read_csv('./Data.csv')
mean_umur = df['Umur'].median();
mean_gaji = df['Gaji'].median();

df['Umur'].fillna(mean_umur, inplace=True)
df['Gaji'].fillna(mean_gaji, inplace=True)

df
```

### 5. Ganti Data Kosong (NaN) pada Umur dan Gaji dengan Nilai Modus

```python
df = pd.read_csv('Data.csv')
mean_umur = df['Umur'].mode()[0];
mean_gaji = df['Gaji'].mode()[0];

df['Umur'].fillna(mean_umur, inplace=True)
df['Gaji'].fillna(mean_gaji, inplace=True)

df
```

### 6. Apabila Ada Umur yang diatas 120 maka ubah menjadi 33

```python
df = pd.read_csv('./Data.csv')

for x in df.index:
    if df.loc[x, 'Umur'] > 120:
        df.loc[x, 'Umur'] = 33

df
```

### 1.Buatlah table list Top 5 Customers yang memiliki nilai transaksi paling banyak !

```python
import pandas as pd

def load_transactions_data():
    """Memuat dan mempersiapkan data transaksi dari file CSV"""
    try:
        # Memuat data transaksi
        transactions_df = pd.read_csv('transactions_1k.csv')

        # Persiapan data transaksi
        transactions_df['DateTime'] = pd.to_datetime(transactions_df['Date'] + ' ' + transactions_df['Time'])
        transactions_df['Hour'] = transactions_df['DateTime'].dt.hour

        return transactions_df
    except FileNotFoundError as e:
        print(f"Error: File tidak ditemukan - {e}")
        return None
    except Exception as e:
        print(f"Error: {e}")
        return None

def analyze_top_transactions(transactions_df):
    """Analisis dan tampilkan 5 transaksi dengan jumlah pembelian terbanyak"""
    if transactions_df is not None:
        print("\n=== 5 TRANSAKSI DENGAN JUMLAH PEMBELIAN TERBANYAK ===")

        # Ambil 5 transaksi dengan Amount terbesar
        top_transactions = transactions_df.nlargest(5, 'Amount')


        print(top_transactions[['TransactionID', 'CustomerID', 'ProductID', 'Amount', 'Price', 'DateTime']].to_string(index=False))

def main():
    """Fungsi utama untuk menjalankan analisis"""


    # Memuat data transaksi
    transactions_df = load_transactions_data()

    # Jalankan analisis
    analyze_top_transactions(transactions_df)



if __name__ == "__main__":
    main()
```

### 2.Buatlah list Top 5 Gas stations yang memiliki nilai transaksi paling banyak !

```python
import pandas as pd

# Membaca file CSV
df_gasstations = pd.read_csv("gasstations.csv")
df_transactions = pd.read_csv("transactions_1k.csv")

# Menghitung total nilai transaksi per stasiun
top_gas_stations = df_transactions.groupby("GasStationID")["Price"].sum().sort_values(ascending=False).head(5).reset_index()

# Menggabungkan dengan informasi stasiun
merged_df = pd.merge(top_gas_stations, df_gasstations, on="GasStationID", how="left")

# Menampilkan hasil
print("\n=== 5 TRANSAKSI DENGAN JUMLAH PEMBELIAN TERBANYAK ===")
print(merged_df)
```

### 3.Buatlah list Top 5 Jenis produk yang memiliki nilai transaksi paling banyak !

```python
import pandas as pd

# Membaca data produk
products = pd.read_csv('products.csv')

# Membaca data transaksi
transactions = pd.read_csv('transactions_1k.csv')

# Menghitung total nilai transaksi (Amount * Price) untuk setiap transaksi
transactions['TotalValue'] = transactions['Amount'] * transactions['Price']

# Menggabungkan data transaksi dengan data produk
merged_data = pd.merge(transactions, products, on='ProductID', how='left')

# Menghitung total nilai transaksi per produk
product_sales = merged_data.groupby(['ProductID', 'Description'])['TotalValue'].sum().reset_index()

# Mengurutkan berdasarkan nilai transaksi terbesar
top_5_products = product_sales.sort_values('TotalValue', ascending=False).head(5)

# Menampilkan hasil
print("Top 5 Jenis Produk dengan Nilai Transaksi Terbanyak:")
print(top_5_products[['Description', 'TotalValue']].to_string(index=False))
```

### 4.Buatlah deskripsi statistik untuk masing masing hari (23,24,25,26) pada dataset transaction_1k !

```python
import pandas as pd
import datetime as dt

# Load semua dataset
df_customer = pd.read_csv('customers.csv')
df_gasstation = pd.read_csv('gasstations.csv')
df_product = pd.read_csv('products.csv')
df_transaction = pd.read_csv('transactions_1k.csv')
df_yearmonth = pd.read_csv('yearmonth.csv')

# Merge semua tabel
df_merged = df_transaction.merge(
    df_customer, on='CustomerID', how='left'
).merge(
    df_gasstation, on='GasStationID', how='left'
).merge(
    df_product, on='ProductID', how='left'
)

# 1. Pastikan kolom Date sudah datetime
df_merged['Date'] = pd.to_datetime(df_merged['Date'])

# 2. Filter tanggal 23-26 Agustus 2012
df_filtered = df_merged[
    df_merged['Date'].isin(pd.to_datetime(['2012-08-23', '2012-08-24', '2012-08-25', '2012-08-26']))
]

# 3. Hitung statistik dengan sintaks yang benar
daily_stats = df_filtered.groupby('Date').agg(
    Amount_count=('Amount', 'count'),
    Amount_mean=('Amount', 'mean'),
    Amount_std=('Amount', 'std'),
    Amount_min=('Amount', 'min'),
    Amount_q1=('Amount', lambda x: x.quantile(0.25)),
    Amount_median=('Amount', 'median'),
    Amount_q3=('Amount', lambda x: x.quantile(0.75)),
    Amount_max=('Amount', 'max'),
    Sales_count=('Price', 'count'),
    Sales_mean=('Price', 'mean'),
    Sales_std=('Price', 'std'),
    Sales_min=('Price', 'min'),
    Sales_q1=('Price', lambda x: x.quantile(0.25)),
    Sales_median=('Price', 'median'),
    Sales_q3=('Price', lambda x: x.quantile(0.75)),
    Sales_max=('Price', 'max')
)

# Tampilkan hasil
daily_stats
```

### 5.Carilah waktu terbaik (hari dan jam) dimana paling banyak user gas station melakukan transaksi !

```python
import pandas as pd
import datetime as dt

# Load semua dataset
df_customer = pd.read_csv('customers.csv')
df_gasstation = pd.read_csv('gasstations.csv')
df_product = pd.read_csv('products.csv')
df_transaction = pd.read_csv('transactions_1k.csv')
df_yearmonth = pd.read_csv('yearmonth.csv')

# Merge semua tabel
df_merged = df_transaction.merge(
    df_customer, on='CustomerID', how='left'
).merge(
    df_gasstation, on='GasStationID', how='left'
).merge(
    df_product, on='ProductID', how='left'
)

# Konversi tipe data waktu
df_merged['Date'] = pd.to_datetime(df_merged['Date'])
df_merged['Hour'] = pd.to_datetime(df_merged['Time']).dt.hour

# Ekstrak jam dari kolom Time
df_merged['Hour'] = pd.to_datetime(df_merged['Time']).dt.hour

# Group by Date dan Hour, lalu hitung total Amount dan Price
best_time = df_merged.groupby(['Date', 'Hour']).agg({
    'Amount': 'sum',
    'Price': 'sum'
}).sort_values('Amount', ascending=False).head(5)

# Rename kolom Price menjadi Gas Sales
best_time = best_time.rename(columns={'Price': 'Gas Sales'})

# Tampilkan hasil
best_time
```
