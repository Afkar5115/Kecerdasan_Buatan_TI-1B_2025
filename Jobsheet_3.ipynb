## Latihan 1

### A. Mendapatkan Nilai Mean

```python
import numpy as np

score = [90, 80, 70, 65, 110, 80, 83, 67, 94]

x = np.mean(score)
print(x)
```

### B. Mendapatkan Nilai Median

```python
import numpy as np

score = [90, 80, 70, 65, 110, 80, 83, 67, 94]

x = np.median(score)
print(x)
```

### C. Mendapatkan Nilai Modus

```python
from scipy import stats

score = [90, 80, 70, 65, 110, 80, 83, 67, 94]

x = stats.mode(score, keepdims=True)
print(x)
```

## Latihan 2

### A. Mendapatkan Nilai Range

```python
import numpy as np

data = [2, 3, 7, 8, 9, 12, 15, 20, 50, 42]

minimum = np.min(data)
maximum = np.max(data)
x = maximum - minimum
print(x)
```

### B. Mendapatkan Nilai Percentile

```python
import numpy as np

data = [90, 80, 70, 65, 110, 80, 83, 67, 94]

x = np.percentile(data, 90)
print(x)
```

### C. Mendapatkan Nilai Quartile dan Interquartile Range

```python
import numpy as np

data = [3, 4, 4, 5, 5, 6, 6, 7, 8, 8, 9, 10]

Q1 = np.median(data[:int(len(data)/2)])
Q3 = np.median(data[int(len(data)/2):])
IQR = Q3 - Q1
print(Q1, Q3, IQR)
```

### D. Mendapatkan Nilai Variance dan Standart Deviation

```python
import numpy as np

data= [86, 87, 88, 86, 87, 85, 86]

mean = np.mean(data)
var = np.var(data)
std = np.std(data)
print(mean, var, std)
```

## Tugas Praktikum 1

```python
import numpy as np
from collections import Counter

data = np.array([1, 2, 3, 4, 5, 5, 6, 2, 3, 4, 5, 6, 6, 4, 3, 2, 1, 4, 3, 5, 6, 6, 5, 4, 5])

frekuensi = Counter(data)

maks_frekuensi = max(frekuensi.values())

modus = np.array([angka for angka, jumlah in frekuensi.items() if jumlah == maks_frekuensi])

print("Modus dari data adalah:", modus)
```

## Tugas Praktikum 3

```python
import numpy as np

# Data hasil ujian
nilai = [50, 54, 62, 50, 52, 59, 61, 63, 65, 10,
         53, 63, 65, 50, 59, 62, 50, 51, 57, 60,
         63, 65, 65, 53, 99]

# a. Urutkan data dari yang terkecil sampai terbesar
nilai_urut = sorted(nilai)
print("a. Data urut:", nilai_urut)

# b. Hitung Q1, Q3, dan IQR
Q1 = np.percentile(nilai, 25)
Q3 = np.percentile(nilai, 75)
IQR = Q3 - Q1

print("b. Q1 (First Quartile):", Q1)
print("   Q3 (Third Quartile):", Q3)
print("   IQR (Interquartile Range):", IQR)

# c. Hitung Lower Bound dan Upper Bound
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("c. Lower Bound:", lower_bound)
print("   Upper Bound:", upper_bound)

# d. Tentukan outlier
outliers = [x for x in nilai if x < lower_bound or x > upper_bound]
print("d. Outlier:", outliers)
```

## Tugas Praktikum 4

```python
# Data distribusi frekuensi berat badan (kelas interval dan frekuensi)
kelas_interval = [(41, 45), (46, 50), (50, 55), (56, 60), (61, 65), (66, 70)]
frekuensi = [2, 3, 6, 4, 5, 2]

# Hitung titik tengah (xi) untuk masing-masing kelas
titik_tengah = [(bawah + atas) / 2 for bawah, atas in kelas_interval]

# Hitung total frekuensi
total_frekuensi = sum(frekuensi)

# Hitung mean (rata-rata) data kelompok
total_fx = sum(f * x for f, x in zip(frekuensi, titik_tengah))
mean = total_fx / total_frekuensi

# Hitung varians dan standar deviasi
sum_f_x2 = sum(f * ((x - mean) ** 2) for f, x in zip(frekuensi, titik_tengah))
varians = sum_f_x2 / total_frekuensi
standar_deviasi = varians ** 0.5

# Cetak data
print("KelasInterval | Frekuensi | Titik Tengah")
for (bawah, atas), f, x in zip(kelas_interval, frekuensi, titik_tengah):
    print(f"{bawah:>2} - {atas:<2}       | {f:^9} | {x:^12}")

print("\nTotal Frekuensi:", total_frekuensi)
print("Mean (Rata-rata):", round(mean, 2))
print("Varians:", round(varians, 2))
print("Standar Deviasi:", round(standar_deviasi, 2))
```
